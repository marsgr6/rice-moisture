```python
from __future__ import print_function
import os
import cv2 
import numpy as np
import pandas as pd   
import matplotlib.pyplot as plt
import argparse
import joblib
```

```python
folders =['C1', 'C2', 'C3', 'C4', 'C5', 'C6', 'C7', 'C8']
folders
```

```python
all_data =[]
DIR = "DATA/TRAINING/"
for folder in folders:
    inFolders = os.listdir(DIR+folder)
    leninFolders = len(inFolders)
    #print(inFolders)
    for infolder in inFolders:
        inFiles = os.listdir(DIR+folder+"/"+infolder)
        lenFiles = len(inFiles)
        all_data.append([folder, leninFolders, infolder, lenFiles])
    #break
```

```python
pdDT = pd.DataFrame(all_data)    # values
pdDT.columns = ["folder", "LenFolders", "Humidity", "Files"]
pdDT
```
